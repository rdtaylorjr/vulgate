complicated patterns with the power of programmatically processing the results.

The ins and outs of installing Text-Fabric, getting the corpus, and initializing a notebook are

```python
from tf.app import use
```

```python
A = use("rdtaylorjr/vulgate", hoist=globals())
```

# Basic search command

We start with the most simple form of issuing a query.
Let's search for the word `pascit` and show the verses in which it occurs.

```python
query = """
verse
  word word=pascit
"""
results = A.search(query)
```

We have the results. We only need to display them. Here they are in a table:

```python
A.table(results)
```

Here is the first one in a pretty display:

```python
A.show(results, end=1)
```

# Condense results

There are two fundamentally different ways of presenting the results: condensed and uncondensed.

In **uncondensed** view, all results are listed individually.
In **condensed** view all nodes of all results are grouped in containers first (e.g. verses), and then presented
container by container.

```python
query = """
verse
  word word=Dominus
"""

results = A.search(query)
```

```python
A.table(results, end=5)
```

Now in condensed mode:

```python
A.table(results, end=5, condensed=True)
```

# Custom highlighting

We can apply different highlight colours to different parts of the result.

```python
query = """
verse
  word word=Dominus
  word word=Deus
"""

results = A.search(query)
```

```python
A.table(results, end=5, condensed=False, colorMap={1: "", 2: "cyan", 3: "magenta"})
```

# Constraining order

You can stipulate an order on the words in your template.
Say we want only results where `Deus` follows `Dominus`.

```python
query = """
verse
  word word=Dominus
  < word word=Deus
"""

results = A.search(query)
```

```python
A.table(results, end=5, colorMap={1: "", 2: "cyan", 3: "magenta"})
```

We can also require the words to be adjacent.

```python
query = """
verse
  word word=Dominus
  <: word word=Deus
"""

results = A.search(query)
```

```python
A.table(results, end=5, colorMap={1: "", 2: "cyan", 3: "magenta"})
```

# Apparatus

The feature `apparatus` holds the critical apparatus. Here we look for words where the Clementine edition (𝔠) alone adds something.

```python
query = """
word apparatus~^\\+.*𝔠\\.$
"""

results = A.search(query)
```

```python
A.table(results, end=5)
```

# Signs

The obelus of Origen is the value `÷` of the feature `prefix`.

```python
query = """
book book=Dn
  word prefix=÷
"""

results = A.search(query)
A.table(results, end=5)
```
